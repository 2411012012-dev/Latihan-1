```python
import requests
import json

print("Memulai pengambilan data...")

url = "https://api.open-meteo.com/v1/forecast"
params = {
    "latitude": -5.45,
    "longitude": 105.26,
    "daily": ["temperature_2m_max", "precipitation_sum"],
    "timezone": "Asia/Jakarta",
    "forecast_days": 7
}

response = requests.get(url, params=params)

if response.status_code == 200:
    data = response.json()

    waktu = data["daily"]["time"]
    suhu = data["daily"]["temperature_2m_max"]
    hujan = data["daily"]["precipitation_sum"]

    with open("hasil_cuaca.txt", "a") as file:
        for i in range(len(waktu)):
            tanggal_hari_ini = waktu[i]
            suhu_hari_ini = suhu[i]
            hujan_hari_ini = hujan[i]

            if suhu_hari_ini > 32.0 or hujan_hari_ini > 0.0:
                data_simpan = {
                    "tanggal": tanggal_hari_ini,
                    "suhu_maks_celsius": suhu_hari_ini,
                    "curah_hujan_mm": hujan_hari_ini
                }

                file.write(json.dumps(data_simpan) + "\n")
                print("[SIMPAN] Data", tanggal_hari_ini, "masuk kriteria. Suhu:", suhu_hari_ini, "C, Hujan:", hujan_hari_ini, "mm")
            else:
                print("[LEWATI] Data", tanggal_hari_ini, "tidak memenuhi kriteria filter.")

    print("Proses selesai. Cek file hasil_cuaca.txt di foldermu.")
else:
    print("Data gagal diambil. Cek koneksi internet.")
```
